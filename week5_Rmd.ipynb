```r
# missing Data
x <- NA
is.na(x)
```

```r
x_vec <- c(11, NA, 13, NA)
is.na(x_vec)
```

```r
x <- c(11, NA, 13, NA)
mean(x)
mean(x, na.rm = TRUE)
x <- c(11, NA, 13, NA)
# loc of na
which(is.na(x))
# count of na
sum(is.na(x))
```

```r
# Finding Complete Cases and Omitting NAs
x <- c(11, NA, 13, NA)
complete.cases(x)
y <- na.omit(x)
y

mean(y)
```

```r
# if condN
x <- 5
if (x > 4) {
  x * 3
}

x <- 6
if (x > 3) {
  print("The value is more than 3")
}
```

```r
#if else
x <- 5
if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}
x


x <- 3
if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}
print(x)



x <- 6
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


```

```r
# nested if else
x <- 5
if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}
x

x <- 3
if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}
x
```

```r
# vectorized if else
x <- 1:10
ifelse(x < 6, x^2, x + 1)
```

```r
x <- c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
```

```r
# return 2 ele
switch(2, "apple", "banana", "orange")
```

```r
x <- c(10, 15, 8, 14, 6, 12)

which(x == 14)
which(x != 12)
which(x > 10)
```

```r
x <- matrix(nrow = 3, ncol = 3, data = 1:9)
x

# min amd max values index
which.min(x)
which.max(x)


# odd numbers
which(x %% 2 == 1)

# array indices (row and col coordinates)
which(x %% 2 == 1, arr.ind = TRUE)
```
